```python
from ingest import load_faq_data

documents = load_faq_data()
```

```python
llm_docs = [doc for doc in documents if doc['course'] == 'llm-zoomcamp']
print(f'LLM Zoomcamp: {len(llm_docs)} documents')
```

```python
from sqlitesearch import TextSearchIndex

index = TextSearchIndex(
    text_fields=['question', 'section', 'answer'],
    keyword_fields=['course'],
    db_path='faq.db'
)
```

```python
import time

for doc in llm_docs:
    index.add(doc)
    time.sleep(0.5)

print('Documents loaded to database!')
```

```python
index.close()
```
